# GameScope Agent — Demonstration

This notebook demonstrates the standalone portfolio package. It validates and indexes the local corpus, runs an evidence-first query, and exposes the workflow trace. Configure `.env` before executing API-dependent cells.

## Setup

```python
from pathlib import Path

from gamescope_agent.config import Settings
from gamescope_agent.data import load_game_records
from gamescope_agent.factory import build_agent
from gamescope_agent.vector_store import ChromaGameStore

settings = Settings.load(Path.cwd().parent)
settings
```

## Validate and index the local corpus

```python
records = load_game_records(settings.data_dir)
store = ChromaGameStore.from_settings(settings)
stored_count = store.ingest(records)
stored_count
```

## Run the evidence-first workflow

```python
agent = build_agent(settings)
run = agent.run(
    "When was Pokémon Gold and Silver released?",
    session_id="notebook-demo",
)
print(run.answer)
run.tool_sequence
```

## Inspect the trace

```python
[event.model_dump() for event in run.trace]
```
